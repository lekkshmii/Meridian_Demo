### Intelligent M&A Target Identification & Analysis

This demonstration showcases Meridian's AI-powered platform that transforms weeks of manual research into actionable intelligence in minutes.

```python
# Initialize Meridian Platform
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Import Meridian modules
import sys
sys.path.append('..')
from meridian import SyntheticDataGenerator, ScreeningEngine, ValuationEngine, IntelligenceAnalyzer
from meridian.visualizations import *

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda x: '%.2f' % x)

print("="*80)
print("                    MERIDIAN M&A INTELLIGENCE PLATFORM")
print("                           Version 1.0.0")
print("="*80)
print("\nInitializing components...")
print("✓ Screening Engine loaded")
print("✓ Valuation Models ready")
print("✓ AI Analysis module active")
print("✓ Connected to data universe")
print("\nSystem ready for analysis.")
```

## 1. Load Target Universe

Access our comprehensive database of potential acquisition targets.

```python
# Load pre-generated universe or create new
import os

if os.path.exists('../data/synthetic_universe.parquet'):
    companies = pd.read_parquet('../data/synthetic_universe.parquet')
    print("Loaded existing universe")
else:
    print("Generating synthetic universe...")
    generator = SyntheticDataGenerator()
    companies = generator.generate_company_universe(5000)
    companies.to_parquet('../data/synthetic_universe.parquet')
    print("Generated and saved universe")

# Initialize screening engine
screener = ScreeningEngine(companies)
analyzer = IntelligenceAnalyzer()

# Display universe statistics
print(f"\nUNIVERSE STATISTICS")
print(f"-" * 40)
print(f"Total Companies: {len(companies):,}")
print(f"Industries: {', '.join(companies['industry'].unique())}")
print(f"Revenue Range: ${companies['revenue_ttm'].min():,.0f} - ${companies['revenue_ttm'].max():,.0f}")
print(f"Average Growth Rate: {companies['revenue_growth_yoy'].mean():.1%}")
print(f"Profitable Companies: {(companies['ebitda'] > 0).sum():,} ({(companies['ebitda'] > 0).mean():.1%})")
```

## 2. Define Acquisition Strategy

Set strategic criteria for target identification based on your investment thesis.

```python
# Define screening criteria
screening_criteria = {
    'min_revenue': 20_000_000,        # $20M minimum
    'max_revenue': 200_000_000,       # $200M maximum
    'min_growth': 0.25,               # 25% minimum growth
    'min_ebitda_margin': -0.05,       # Can be slightly unprofitable
    'industries': ['SaaS', 'Fintech', 'Data & Analytics'],
    'max_debt_ratio': 1.5,            # Max 1.5x debt/equity
    'require_profitability': False,   # Include growth-stage companies
    'exclude_distressed': True        # Exclude distressed assets
}

print("ACQUISITION CRITERIA")
print("-" * 40)
print(f"Target Revenue: ${screening_criteria['min_revenue']/1e6:.0f}M - ${screening_criteria['max_revenue']/1e6:.0f}M")
print(f"Minimum Growth: {screening_criteria['min_growth']:.0%}")
print(f"Industries: {', '.join(screening_criteria['industries'])}")
print(f"Profitability Required: {screening_criteria['require_profitability']}")
print(f"\nStrategy: Focus on high-growth technology companies with proven product-market fit")
```

## 3. Execute Intelligent Screening

Apply multi-dimensional screening with AI-powered scoring.

```python
# Execute screening
print("🔍 Executing screening analysis...\n")
results = screener.screen(**screening_criteria)

# Get screening summary
summary = screener.get_screening_summary()

print("SCREENING COMPLETE")
print("=" * 50)
print(f"Targets Identified: {summary['total_targets']}")
print(f"Immediate Action: {summary['immediate_action']} companies")
print(f"Priority Targets: {summary['priority']} companies")
print(f"Average Score: {summary['average_score']:.1f}/100")
print(f"\n Top Target: {summary['top_target']}")

# Show screening funnel
funnel_data = screener.screening_history[-1]['funnel']
fig = create_screening_funnel(funnel_data)
fig.show()
```

## 4. Top Targets Analysis

Detailed view of the highest-scoring acquisition opportunities.

```python
# Display top 10 targets
top_10 = results.head(10)

# Create strategic positioning chart
fig = create_target_scatter(
    top_10,
    title="Top 10 M&A Targets - Strategic Positioning"
)
fig.show()

# Display summary table
display_columns = [
    'rank', 'name', 'industry', 'sub_industry',
    'revenue_ttm', 'revenue_growth_yoy', 'ebitda_margin',
    'implied_revenue_multiple', 'total_score', 'category'
]

display_df = top_10[display_columns].copy()
display_df['revenue_ttm'] = display_df['revenue_ttm'].apply(lambda x: f"${x/1e6:.1f}M")
display_df['revenue_growth_yoy'] = display_df['revenue_growth_yoy'].apply(lambda x: f"{x:.1%}")
display_df['ebitda_margin'] = display_df['ebitda_margin'].apply(lambda x: f"{x:.1%}")
display_df['implied_revenue_multiple'] = display_df['implied_revenue_multiple'].apply(lambda x: f"{x:.1f}x")
display_df['total_score'] = display_df['total_score'].apply(lambda x: f"{x:.1f}")

from IPython.display import HTML
html = display_df.to_html(index=False, classes='table table-striped')
display(HTML(html))
```

## 5. Deep Dive: Primary Target

Comprehensive analysis of the top-scoring acquisition candidate.

```python
# Select top target for deep analysis
top_target = results.iloc[0]

print(" PRIMARY ACQUISITION TARGET")
print("="*60)
print(f"Company: {top_target['name']}")
print(f"Industry: {top_target['industry']} - {top_target['sub_industry']}")
print(f"Headquarters: {top_target['headquarters']}")
print(f"Founded: {top_target['founded_year']}")
print(f"Employees: {top_target['employees']:,}")
print(f"\n FINANCIAL METRICS")
print("-"*40)
print(f"Revenue (TTM): ${top_target['revenue_ttm']:,.0f}")
print(f"Growth Rate: {top_target['revenue_growth_yoy']:.1%}")
print(f"EBITDA Margin: {top_target['ebitda_margin']:.1%}")
print(f"Valuation: {top_target['implied_revenue_multiple']:.1f}x revenue")
print(f"Enterprise Value: ${top_target['enterprise_value']:,.0f}")
```

## 6. AI-Powered Strategic Fit Analysis

Leverage artificial intelligence to assess strategic alignment and synergy potential.

```python
# Simulate acquirer profile
acquirer_profile = {
    'name': 'Strategic Acquirer Corp',
    'industry': 'SaaS',
    'revenue': 500_000_000,
    'founded_year': 2010,
    'employees': 2000
}

print(" Analyzing strategic fit using AI...\n")

# Perform AI analysis
strategic_analysis = analyzer.analyze_strategic_fit(
    acquirer_profile,
    top_target.to_dict(),
    analysis_depth='comprehensive'
)

# Display strategic fit scores
fig = create_strategic_fit_radar(
    strategic_analysis['scores'],
    top_target['name']
)
fig.show()

print("\n KEY INSIGHTS:")
for i, insight in enumerate(strategic_analysis['insights'], 1):
    print(f"{i}. {insight}")

print(f"\nOVERALL STRATEGIC FIT: {strategic_analysis['overall_fit']:.1f}/100")
print(f" RECOMMENDATION: {strategic_analysis['recommendation']}")
print(f" CONFIDENCE LEVEL: {strategic_analysis['confidence']:.1%}")
```

## 7. Valuation Analysis

Multi-methodology valuation to determine fair value and potential returns.

```python
# Initialize valuation engine
valuation_engine = ValuationEngine()

# Load comparables (if available)
industry_safe = top_target['industry'].replace(' ', '_').replace('&', 'and')
comp_file = f'../data/comparables_{industry_safe}.parquet'

if os.path.exists(comp_file):
    comparables = pd.read_parquet(comp_file)
else:
    # Generate comparables
    generator = SyntheticDataGenerator()
    comparables = generator.generate_market_comparables(top_target['industry'], 30)

# Perform comprehensive valuation
valuation_results = valuation_engine.comprehensive_valuation(
    top_target,
    comparables=comparables
)

print(" VALUATION ANALYSIS")
print("="*60)
print(f"Current Enterprise Value: ${valuation_results['current_ev']:,.0f}")
print("\nValuation by Methodology:")
print("-"*40)

for method, result in valuation_results['methodologies'].items():
    if 'value' in result:
        print(f"{method.upper()}: ${result['value']:,.0f}")

if 'valuation_range' in valuation_results:
    print("\n VALUATION SUMMARY:")
    print(f"Min Value: ${valuation_results['valuation_range']['min']:,.0f}")
    print(f"Max Value: ${valuation_results['valuation_range']['max']:,.0f}")
    print(f"Median Value: ${valuation_results['valuation_range']['median']:,.0f}")
    print(f"\n IMPLIED RETURN: {valuation_results['implied_return']['median']:.1%}")

# Create valuation football field
fig = create_valuation_football_field(
    valuation_results['methodologies'],
    top_target['enterprise_value'],
    top_target['name']
)
fig.show()
```
